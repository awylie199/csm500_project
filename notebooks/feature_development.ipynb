- All data across datasets has been aggregated, cleaned and consolidated.
- Superfluous data has been excised (such as ID columns which offer no explanatory value).
- Sparse data has been removed where necessary (such as description or addresses) due to their essential explanatory qualities.
- Raw geospatial data in the primary application dataset has been used to add the Index of Multiple Deprivation and the corresponding decile, LSOA, the county and whether the application is within a conservation area.
- Further supplementary geospatial data has been added after extensive refinement of government datasets
- Anomalous applications which fall within unknown LSOAs (ostensibly due to data entry errors) were removed due to suspected corruption.
- Date features were disaggregated into appropriate dimensions (ignoring more granular time-units).

The decision to remove anomalous or sparse results may introduce selection bias, but are a very small minority of the overall dataset.

```python
from sklearn.model_selection import train_test_split

TRAIN_TEST_RANDOM_STATE: Final[int] = 20260902

x_train_final, x_test, y_train_final, y_test = train_test_split(
    gdf.drop(columns=[INFERRED_TARGET_FIELD]),
    gdf[[INFERRED_TARGET_FIELD]],
    test_size=0.33,
    random_state=TRAIN_TEST_RANDOM_STATE,
    stratify=gdf[[INFERRED_TARGET_FIELD]]
)

# Create copies for univariate and covariate analysis
x_train = x_train_final.copy(deep=True)
y_train = y_train_final.copy(deep=True)

x_train.reset_index(drop=True)
y_train.reset_index(drop=True)

x_train_final.to_parquet(RAW_TRAIN_DATASET_X_PATH, engine="pyarrow", compression="snappy")
y_train_final.to_parquet(RAW_TRAIN_DATASET_Y_PATH, engine="pyarrow", compression="snappy")
x_test.to_parquet(RAW_TEST_DATASET_X_PATH, engine="pyarrow", compression="snappy")
y_test.to_parquet(RAW_TEST_DATASET_Y_PATH, engine="pyarrow", compression="snappy")

print("Saved Train/Test Datasets to Cache.")
```

The dataset is composed mostly of discrete categorical data and textual data, with one sparse continuous feature,.

Before subsequent analysis, to prevent leakage, the dataset is split into train and test datasets.

A copy of the split train dataset will be used for further analysis and refinement.

```python
# Reassign floor area as transformed with box-cox for Gaussian-like distribution
x_train["Floor Area (m^2)"] = continuous_data_analysis(x_train.copy(deep=True))
```

The dataset contains one continuous feature which is positive and unbounded in nature as it represents the floor area of the property in square metres - it is also a highly sparse feature, retrieved from a joined dataset for the environmental efficiency of the property via another dataset that facilitates matching the UPRN by location.

Normaltest and Jarque-Bera were employed to evaluate the distribution by skewness and kurtosis, for comparison to a Gaussian distribution which will enable a wider selection of downstream models. These tests were chosen due to their robustness with a high sample count.

Although both tests fail to reject the null hypothesis after transformation (and therefore, it is not a normal distribution), visual inspection shows that for the purposes of this project, the distribution is satisfactory: the values only diverge at the outliers. Skewness is close to zero (-0.14), but the heavily left-tailed distribution leads to slightly abnormal kurtosis of 1.92 (lower than expected 3.0 for a perfect Gaussian distribution).

```python
text_transform_dfs = text_data_analysis(x_train.copy(deep=True), y_train.copy(deep=True))
```

```python
(x_train_primary_with_pf_df, primary_mutual_class_info_scores, y_train_primary), (x_train_physical_with_pf_df,
                                                                                  physical_mutual_class_info_scores,
                                                                                  y_train_physical) = correlation_analysis(
    x_train, y_train,
    text_transform_dfs)
```

To better facilitate correlation analysis, the dataset was split into the primary and physical dataset (which is a subset of the primary dataset, but with all physical features).

The same vocabulary was found between text encoding methods.

Significant intercorrelation was found between expected features, such as geological hazards and administrative regions - i.e. if an area is susceptible to a specific geohazard area, it is liable to others, similarly parishes, LSOAs, councils share geographical proximity. Similarly, as expected, IMD rank and IMD decile overlap. Pearson's correlation was generally stronger across the matrix. Notably, Council 32 exhibits almost perfect multicollinearity with several features.

As the target is multi-class, Pearson's and Spearman's were not employed for measuring correlation with the target - only to explore multicollinearity between explanatory variables.

Very little correlation was found between features and the target. However, more correlation was found between the target and the text encodings. Further comparison will be done between the encoding and embedding methods to select the best.

Features with multicollinearity, or very little correlation, were dropped to reduce dimensionality.

Notably, more correlation was observed with the target for the physical dataset, with different features ranking as the most correlated, implying significant  for properties that have and energy efficiency certificate (and thereby extracted physical properties). Nevertheless, the top 15 are broadly similar between datasets.

Within the physical dataset, the absence of correlation between features (due to the absence of values) would imply that specific administrative regions have not added records to the underlying government dataset.

```python
# Drop columns with less than 1% mutual info class score from both datasets, and multicollinearity columns
drop_primary_columns = set(primary_mutual_class_info_scores[primary_mutual_class_info_scores < 0.01].index.tolist())
drop_physical_columns = set(physical_mutual_class_info_scores[physical_mutual_class_info_scores < 0.01].index.tolist())

multicollinearity_columns: set[str] = {"IMD Rank",
                                       "Council Code_E10000032"}

x_train_primary_with_pf_df.drop(columns=drop_primary_columns.union(multicollinearity_columns), inplace=True)
x_train_physical_with_pf_df.drop(columns=drop_physical_columns.union(multicollinearity_columns), inplace=True)

print(x_train_primary_with_pf_df.shape)
print(x_train_physical_with_pf_df.shape)
```
